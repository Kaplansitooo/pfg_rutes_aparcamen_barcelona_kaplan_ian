## Part 0 - Setup

```python
import pandas as pd 
import os
import geopandas as gpd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine, text
from dotenv import load_dotenv
from sqlalchemy.engine import URL


load_dotenv(override=True)
url = URL.create(
    "postgresql+psycopg2",
    username=os.getenv("POSTGRES_USER"),
    password=os.getenv("POSTGRES_PASSWORD"),
    host="127.0.0.1",
    port=5432,
    database="explore",
)
engine = create_engine(url)

def q(sql):
    return pd.read_sql(text(sql), engine)
    
```

```python
q("SELECT postgis_version();")
```

```python
print(q("SELECT table_name FROM information_schema.tables WHERE table_schema = 'public' ORDER BY table_name;"))
print(q("SELECT extname FROM pg_extension ORDER BY extname"))
```

```python
import shutil
from pathlib import Path 
PBF = Path("../data/osm/barcelona.osm.pbf")
OMS2PGSQL = shutil.which("osm2pgsql")

if not PBF.exists():
    raise shutil.Error(f"Missing OSM File: {PBF}")
if OMS2PGSQL is None:
    raise shutil.Error("osm2pgsql not found in Path")

print(f"OSM file: {PBF}")
print(f"osm2pgsql: {OMS2PGSQL}")
```
